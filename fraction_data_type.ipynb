```python
class Fraction:
    def __init__(self,x,y):
        self.num=x
        self.denum=y
    def __str__(self):
        return '{}/{}'.format(self.num,self.denum)
    def __add__(self,other):
        new_num=self.num*other.denum+self.denum*other.num
        new_denum=self.denum*other.denum
        return '{}/{}'.format(new_num,new_denum)
    def __sub__(self,other):
        new_num=self.num*other.denum-self.denum*other.num
        new_denum=self.denum*other.denum
        return '{}/{}'.format(new_num,new_denum)
    def __mul__(self,other):
        new_num=self.num*other.num
        new_denum=self.denum*other.denum
        return '{}/{}'.format(new_num,new_denum)
    def __truediv__(self,other):
        new_num=self.num*other.denum
        new_denum=self.denum*other.num
        return '{}/{}'.format(new_num,new_denum)
obj=Fraction(3,4)
obj2=Fraction(1,2)
print(obj-obj2)
print(obj+obj2)
print(obj*obj2)
print(obj/obj2)
```
